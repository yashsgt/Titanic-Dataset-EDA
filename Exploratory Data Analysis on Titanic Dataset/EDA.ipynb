## Steps for EDA:

#### Categorise the given column as a Numerical, Categorical, Mixed Data type column
#### Univariate Analysis --> Numerical Column or Categorical Column (Analysis over a single column)
#### Numerical Column Analysis --> Distribution of the data --> Histogram, kdeplot
#### Outliers detection --> (BoxPlot)
#### Handling Missing Values 

#### Categorical Column Analysis --> (Univariate Analysis) --> Distribution of Data --> countplot --> Missing Values

#### Bivariate Analysis --> Numerical-Numerical column --> Scatterplot
#### Numerical-Categorical --> barplot
#### Categorical-Categorical --> Heatmap

#### EDA --> Feature Engineering

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df
```

# Univariate Analysis of Numerical Column

```python
df['Age'].describe()
```

```python
df[df['Age']==80]
```

```python
sns.boxplot(data=df, x='Age')
plt.show()
```

```python
df['Age'].isnull().sum()
```

```python
df['Age'].median()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
df['Age'].describe()
```

```python
sns.histplot(data=df, x='Age')
```

```python
sns.kdeplot(data=df, x='Age')
plt.show()
```

# Univariate Analysis of Categorical Column

```python
df.head()
```

```python
df['Survived'].value_counts()
```

```python
sns.countplot(data=df, x='Survived')
```

```python
df['Survived'].isnull().sum()   # No Missing Value
```

```python
df['Embarked'].value_counts()
```

```python
sns.countplot(data=df, x='Embarked')
```

```python
df['Embarked'].isnull().sum()
```

```python
df['Embarked'].mode()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Embarked'].isnull().sum()
```

# Bivariate Analysis
### 1. Numerical-Numerical Analysis
### 2. Numerical-Categorical Analysis
### 3. Categorical-Categorical Analysis

```python
## Numerical-Numerical Analysis

sns.scatterplot(data=df, x='Age', y='Fare')
plt.show()
```

```python
# Numerical-Categorical Analysis:

sns.barplot(data=df, x='Pclass', y='Fare')
plt.show()
```

```python
# Numerical-Categorical Analysis:

sns.barplot(data=df, x='Embarked', y='Fare')
plt.show()
```

```python
df['SibSp'].value_counts()
```

```python
# Numerical-Categorical Analysis:
sns.barplot(data=df, x='SibSp', y='Fare')
plt.show()
```

```python
# Numerical-Categorical Analysis:
sns.barplot(data=df, x='Sex', y='Fare')
plt.show()
```

```python
## Categorical-Categorical Analysis

## We will use crosstab() --> two categories as input on which we can perform the analysis
```

```python
pd.crosstab(df['Pclass'], df['Survived'], normalize=True)*100
## Passenger of Pclass 1 survived the most as compared to other Pclass
```

```python
sns.heatmap(pd.crosstab(df['Pclass'], df['Survived'], normalize=True)*100)
```

```python
pd.crosstab(df['Sex'], df['Survived'], normalize=True)*100
```

```python
pd.crosstab(df['Embarked'], df['Survived'], normalize=True)*100
```
